```python

# Load SPX 5-minute data from local file
def load_spx_data(file_path, months_back=3):
    """Load SPX data and filter for last N months"""
    # Read the CSV file
    df = pd.read_csv(file_path, 
                     names=['datetime', 'open', 'high', 'low', 'close'],
                     parse_dates=['datetime'])
    
    # Calculate the date cutoff (3 months back from latest date)
    latest_date = df['datetime'].max()
    cutoff_date = latest_date - timedelta(days=months_back * 30)
    
    # Filter for last 3 months
    df_filtered = df[df['datetime'] >= cutoff_date].copy()
    df_filtered = df_filtered.set_index('datetime')
    
    print(f"Loaded {len(df_filtered)} records from {df_filtered.index.min()} to {df_filtered.index.max()}")
    return df_filtered

# Load the data (using market_data symlink accessible from notebooks)
spx_data = load_spx_data('market_data/historical/SPX/5M.txt', months_back=3)
print(f"SPX data shape: {spx_data.shape}")
print(f"Columns: {list(spx_data.columns)}")
spx_data.head()
```
